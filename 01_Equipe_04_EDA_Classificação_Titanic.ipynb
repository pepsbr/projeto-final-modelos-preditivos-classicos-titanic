## Normalização dos dados

Hora de para padronizar as variáveis.

Import StandardScaler do Scikit-learn.

Converta as variáveis independentes com o StandardScaler;

Use o método .transform () para transformar os parâmetros em uma versão padronizada.

####Normalização do Fare e Age

```python
from sklearn.preprocessing import StandardScaler

fare_scaler = StandardScaler()
age_scaler = StandardScaler()

# Para os dados de Treino
X_train_fare = X_train[["Fare"]]
X_train_age = X_train[["Age"]]

X_train_fare_scaled = fare_scaler.fit_transform(X_train[["Fare"]])
X_train_age_scaled = age_scaler.fit_transform(X_train[["Age"]])

# Para dados de Teste
X_test_fare = X_test[["Fare"]]
X_test_age = X_test[["Age"]]

X_test_fare_scaled = fare_scaler.transform(X_test[["Fare"]])
X_test_age_scaled = age_scaler.transform(X_test[["Age"]])

print('Dados Treino do Fare')
display(X_train_fare_scaled[:10])
print('\nDados do Treino do Age')
display(X_train_age_scaled[:10])

print('\nDados Teste do Fare')
display(X_test_fare_scaled[:10])
print('\nDados do Teste do Age')
display(X_test_age_scaled[:10])
```

```python
# Criação do dataframes temporarios para as informações de treino escaladas de Fare e Age
db_x_train_fare = pd.DataFrame(X_train_fare_scaled, index=X_train.index)
db_x_train_fare = db_x_train_fare.rename(columns={0: 'Fare_scaled'})

db_x_train_age = pd.DataFrame(X_train_age_scaled, index=X_train.index)
db_x_train_age = db_x_train_age.rename(columns={0: 'Age_scaled'})

# Concatenação no X_train_scaled, as informações das colunas scaled
X_train_scaled = X_train.copy()
X_train_scaled = pd.concat([X_train_scaled, db_x_train_fare], axis = 1)
X_train_scaled = pd.concat([X_train_scaled, db_x_train_age], axis = 1)

# Criação do dataframes temporarios para as informações de teste escaladas de Fare e Age
db_x_test_fare = pd.DataFrame(X_test_fare_scaled, index=X_test.index)
db_x_test_fare = db_x_test_fare.rename(columns={0: 'Fare_scaled'})

db_x_test_age = pd.DataFrame(X_test_age_scaled, index=X_test.index)
db_x_test_age = db_x_test_age.rename(columns={0: 'Age_scaled'})

# Concatenação no X_test_scaled, as informações das colunas scaled
X_test_scaled = X_test.copy()
X_test_scaled = pd.concat([X_test_scaled, db_x_test_fare], axis = 1)
X_test_scaled = pd.concat([X_test_scaled, db_x_test_age], axis = 1)
```

## Conclusão

Neste estudo, aplicamos ao dataset do Titanic (891 passageiros, 12 colunas) as etapas de pré-processamento necessárias para deixar os dados prontos para um modelo de classificação da variável `Survived`.

**O que foi feito**
- **Análise inicial:** `info()` e `describe()` mostraram valores ausentes em `Age` (177), `Cabin` (687, cerca de 77% da coluna) e `Embarked` (2). Também mostraram que `Fare` é bastante assimétrica (média 32,2; mediana 14,5; máximo 512,3). Cerca de 38% dos passageiros sobreviveram.
- **Transformação de categóricas:** `Sex` foi codificada com `LabelEncoder` (`Sex_le`) e `Embarked` foi binarizada com `get_dummies` (`Embarked_Q`, `Embarked_S`, tendo C como categoria de referência).
- **Remoção de campos:** `PassengerId`, `Name`, `Ticket` e `Cabin` foram descartados por serem identificadores ou terem excesso de dados faltantes. `Sex` também foi removida após a codificação.
- **Valores nulos:** as idades ausentes foram preenchidas com a mediana de cada grupo (classe × sexo), preservando mais a estrutura dos dados do que uma mediana global.
- **Separação holdout:** os dados foram divididos em 70% para treino (623 linhas) e 30% para teste (268 linhas), com `random_state=111` para reprodutibilidade.
- **Normalização:** `Age` e `Fare` foram padronizadas com `StandardScaler` (média 0 e desvio 1). As demais variáveis foram mantidas, pois são binárias ou categorias/contagens com escala pequena.

**Por que separar treino e teste**
O conjunto de teste simula dados que o modelo nunca viu. Ele serve para estimar o desempenho real do modelo, e essa estimativa só é confiável se o teste não influenciar nenhuma etapa do treinamento.

**Por que o scaler foi ajustado (`fit`) só no treino**
O `StandardScaler` aprende a média e o desvio padrão de cada coluna, então faz parte do "conhecimento" do modelo. Por isso usamos `fit_transform` no treino e apenas `transform` no teste, aplicando a mesma régua aprendida no treino. Se o `fit` usasse também o teste, informação dos dados de avaliação vazaria para o pré-processamento (*data leakage*) e a métrica ficaria otimista. Em um cenário real, um passageiro novo só poderia ser escalado com as estatísticas guardadas do treino.

O resultado confirma o comportamento esperado: no treino, as variáveis escaladas têm média ≈ 0 e desvio ≈ 1; no teste, ficam próximas, mas não exatas (por exemplo, desvio de 1,25 em `Fare_scaled` e de 0,92 em `Age_scaled`). Isso é normal e indica que o teste foi tratado como dado novo. O `inverse_transform` permitiu recuperar os valores originais a partir dos escalados, validando a transformação.

**Limitações**
- `Fare` continua assimétrica após a padronização.
